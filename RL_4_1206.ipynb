## Multi-Armed Bandit with Upper Confidence Bound (UCB)

This notebook implements the UCB algorithm to solve a Multi-Armed Bandit problem with three slot machines (A, B, C). The objective is to estimate the future reward (Q-value) of each machine using the incremental sample-average method over 10 plays.

```python
import numpy as np

machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10

N = np.zeros(M, dtype=int)
R = np.zeros(M, dtype=float)
Q = np.zeros(M, dtype=float)
UCB = np.zeros(M, dtype=float)

print(f"Machines: {machines}")
print(f"Number of plays: {N_play}")
print("\nInitial state:")
print(f"N (Play counts): {N}")
print(f"R (Total rewards): {R}")
print(f"Q (Estimated values): {Q}")
print(f"UCB (Upper Confidence Bound): {UCB}")
```

### Main Simulation Loop

```python
print("\nStarting simulation...")

for t in range(1, N_play + 1):
    print(f"\n--- Play {t} ---")

    for i in range(M):
        if N[i] == 0:
            UCB[i] = np.inf
        else:
            UCB[i] = Q[i] + np.sqrt(2 * np.log(t) / N[i])

    print(f"UCB values for this round: {UCB}")

    selected_machine_index = np.argmax(UCB)
    selected_machine_name = machines[selected_machine_index]
    print(f"Selected machine: {selected_machine_name} (index {selected_machine_index})")

    while True:
        try:
            reward_input = float(input(f"Play with machine {selected_machine_name}. Enter reward: "))
            break
        except ValueError:
            print("Invalid input. Please enter a numerical reward.")

    N[selected_machine_index] += 1
    R[selected_machine_index] += reward_input

    Q[selected_machine_index] = R[selected_machine_index] / N[selected_machine_index]

    print(f"After play {t} (machine {selected_machine_name}, reward {reward_input}):")
    print(f"  N (Play counts): {N}")
    print(f"  R (Total rewards): {R}")
    print(f"  Q (Estimated values): {Q}")
```

### Part F – Display Results

```python
print("\n--- Final Results after all plays ---")
print(f"Total plays: {N_play}")

print("\nMachine Performance Summary:")
for i, machine in enumerate(machines):
    print(f"Machine {machine}:")
    print(f"  Times played (N): {N[i]}")
    print(f"  Total reward (R): {R[i]:.2f}")
    print(f"  Estimated Q-value: {Q[i]:.2f}")

print("\nFinal Tables:")
print(f"N (Play counts): {N}")
print(f"R (Total rewards): {R}")
print(f"Q (Estimated values): {Q}")

final_ucb = np.zeros(M, dtype=float)
for i in range(M):
    if N[i] == 0:
        final_ucb[i] = np.inf
    else:
        final_ucb[i] = Q[i] + np.sqrt(2 * np.log(N_play) / N[i])

print(f"UCB (Upper Confidence Bound): {final_ucb}")

total_reward_player = np.sum(R)
print(f"\nTotal reward the player received: {total_reward_player:.2f}")
```

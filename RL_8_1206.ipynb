```python

GAMMA = 0.9

NORMAL_REWARD = -1
GOAL_REWARD = 10

policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}
transitions = {
    ("S0", "Right"): ("S1", -1),
    ("S0", "Down"): ("S0", -1),
    ("S0", "Left"): ("S0", -1),
    ("S0", "Up"): ("S0", -1),

    ("S1", "Right"): ("G", 10),
    ("S1", "Down"): ("S3", -1),
    ("S1", "Left"): ("S0", -1),
    ("S1", "Up"): ("S1", -1),

    ("S3", "Right"): ("S4", -1),
    ("S3", "Left"): ("X", -5),
    ("S3", "Up"): ("S1", -1),
    ("S3", "Down"): ("S3", -1),

    ("S4", "Right"): ("S4", -1),
    ("S4", "Left"): ("S3", -1),
    ("S4", "Up"): ("G", 10),
    ("S4", "Down"): ("S4", -1)
}

def state_value_function(state):


    total_return = 0
    discount = 1

    current_state = state

    while current_state != "G":
        action = policy[current_state]

        next_state, reward = transitions[(current_state, action)]

        total_return += discount * reward

        discount *= GAMMA

        current_state = next_state

    return total_return

def action_value_function(state, action):


    total_return = 0
    discount = 1

    current_state = state
    current_action = action

    while current_state != "G":

        next_state, reward = transitions[(current_state, current_action)]

        total_return += discount * reward

        discount *= GAMMA

        current_state = next_state

        if current_state != "G":
            current_action = policy[current_state]

    return total_return



print("STATE VALUES")

for state in ["S0", "S1", "S3", "S4"]:
    value = state_value_function(state)
    print(f"V({state}) = {value:.4f}")


print("\nACTION VALUE")

q_value = action_value_function("S0", "Up")

print("Q(S0, Up) =", round(q_value, 4))
```

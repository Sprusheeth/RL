# Assignment 1: Algebraic Bellman Expectation Equations

```python
gamma = 0.9

V0 = (-1 + 0.81 * 89.9) / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print(f"Vπ(S0) = {V0:.2f}")
print(f"Vπ(S1) = {V1:.2f}")
print(f"Qπ(S0, Down) = {Q_S0_Down:.2f}")
print(f"Qπ(S0, Right) = {Q_S0_Right:.2f}")
print(f"Qπ(S1, Right) = {Q_S1_Right:.2f}")
print(f"Qπ(S1, Up) = {Q_S1_Up:.2f}")
```

# Assignment 2: Iterative Policy Evaluation

```python
gamma = 0.9
theta = 0.000001
V0 = 0.0
V1 = 0.0
iterations = 0

while True:
    new_V0 = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
    new_V1 = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))
    V0, V1 = new_V0, new_V1
    iterations += 1
    if delta < theta:
        break

Q_S0_Down = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
Q_S0_Right = -1
Q_S1_Right = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
Q_S1_Up = -1 + gamma * V0

print(f"Iterations = {iterations}")
print(f"Vπ(S0) = {V0:.6f}")
print(f"Vπ(S1) = {V1:.6f}")
print(f"Qπ(S0, Down) = {Q_S0_Down:.6f}")
print(f"Qπ(S0, Right) = {Q_S0_Right:.6f}")
print(f"Qπ(S1, Right) = {Q_S1_Right:.6f}")
print(f"Qπ(S1, Up) = {Q_S1_Up:.6f}")
print(f"Policy check S0: {abs(V0 - Q_S0_Down) < theta}")
print(f"Policy check S1: {abs(V1 - Q_S1_Right) < theta}")
```

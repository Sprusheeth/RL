```python
import numpy as np

machines_rewards = {}

num_machines = int(input("Enter the number of machines: "))

for i in range(num_machines):
    machine_name = input(f"Enter the name for machine {i+1}: ")
    rewards_input = input(f"Enter comma-separated rewards for machine {machine_name} ")
    rewards_input_processed = rewards_input.replace(' ', ',')
    rewards_list = [int(x.strip()) for x in rewards_input_processed.split(',') if x.strip()]
    machines_rewards[machine_name] = rewards_list

estimated_q_values = {}

for machine, rewards in machines_rewards.items():
    q_a = np.mean(rewards)
    estimated_q_values[machine] = q_a
    print(f"Machine {machine}")
    print(f"Rewards: {rewards}")
    print(f"Q({machine}) = {q_a:.2f}")
    print()

greedy_selected_machine = max(estimated_q_values, key=estimated_q_values.get)

print(f"Greedy Selected Machine: {greedy_selected_machine}")
```

```python
import numpy as np

machines_rewards_input = {}

num_machines = int(input("Enter the number of machines: "))

for i in range(num_machines):
    machine_name = input(f"Enter the name for machine {i+1}: ")
    rewards_input = input(f"Enter comma-separated rewards for machine {machine_name} (e.g., 3,2,4): ")
    rewards_input_processed = rewards_input.replace(' ', ',')
    rewards_list = [int(x.strip()) for x in rewards_input_processed.split(',') if x.strip()]
    machines_rewards_input[machine_name] = rewards_list

q_values = {name: 0.0 for name in machines_rewards_input.keys()}
counts = {name: 0 for name in machines_rewards_input.keys()}


for machine_name, all_rewards_observed in machines_rewards_input.items():
    print(f"\nProcessing rewards for Machine {machine_name} (Total: {all_rewards_observed})")
    for reward in all_rewards_observed:
        counts[machine_name] += 1
        n_a = counts[machine_name]

        q_old_a = q_values[machine_name]
        q_new_a = q_old_a + (reward - q_old_a) / n_a
        q_values[machine_name] = q_new_a

        print(f"  - Observed Reward: {reward}, N({machine_name}): {n_a}, Q({machine_name}) updated to {q_new_a:.2f}")

for machine, q_val in q_values.items():
    print(f"Machine {machine}")
    print(f"Q({machine}) = {q_val:.2f}")
    print(f"N({machine}) = {counts[machine]}")
    print()

greedy_selected_machine = max(q_values, key=q_values.get)

print(f"Greedy Selected Machine: {greedy_selected_machine}")
```

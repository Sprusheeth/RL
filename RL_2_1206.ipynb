```python
import random

machines = ['A', 'B', 'C']

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

N_play = 10
epsilon = 0.2

print("Epsilon-Greedy Strategy for Multi-Armed Bandit")
print("Machines:", machines)
print("Number of Plays:", N_play)
print("Epsilon:", epsilon)

for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:
        max_q = max(Q.values())
        best_machines = [machine for machine in machines if Q[machine] == max_q]
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    print("Updated Q-values:")
    print("A =", round(Q['A'], 2),
          "B =", round(Q['B'], 2),
          "C =", round(Q['C'], 2))

print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nN Table - Participation Count")
for machine in machines:
    print(machine, ":", N[machine])

print("\nR Table - Total Rewards")
for machine in machines:
    print(machine, ":", R[machine])

print("\nQ Table - Estimated Values")
for machine in machines:
    print(machine, ":", round(Q[machine], 2))

print("\nAnalysis")
for machine in machines:
    if N[machine] > 0:
        print(
            "Machine", machine,
            "was selected", N[machine],
            "times, received total reward", R[machine],
            "and has estimated Q-value", round(Q[machine], 2)
        )
    else:
        print("Machine", machine, "was never selected.")
```

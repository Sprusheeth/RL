```python
machines = ['A', 'B', 'C']
M = len(machines)

N_play = 10
epsilon = 0.2

N = {machine: 0 for machine in machines}

Q = {machine: 0.0 for machine in machines}

print(f"Machines: {machines}")
print(f"Number of Plays: {N_play}")
print(f"Exploration Probability (epsilon): {epsilon}")
print(f"Initial N (play counts): {N}")
print(f"Initial Q (estimated rewards): {Q}")
```

### Main Simulation Loop (Part C, D, E)

```python
import random

total_reward_collected = 0

for play in range(N_play):
    print(f"\n--- Play {play + 1}/{N_play} ---")

    r = random.random()
    if r < epsilon:
        print("Performing exploration...")
        max_q_machine = max(Q, key=Q.get)

        other_machines = [m for m in machines if m != max_q_machine]
        if other_machines:
            selected_machine = random.choice(other_machines)
        else:
            selected_machine = max_q_machine

    else:
        print("Performing exploitation...")
        selected_machine = max(Q, key=Q.get)

    print(f"Selected machine: {selected_machine}")

    while True:
        try:
            reward_input = input(f"Enter reward obtained from machine {selected_machine}: ")
            current_reward = float(reward_input)
            break
        except ValueError:
            print("Invalid input. Please enter a numerical reward.")

    total_reward_collected += current_reward

    N[selected_machine] += 1

    if N[selected_machine] > 0:
        Q[selected_machine] = Q[selected_machine] + (1 / N[selected_machine]) * (current_reward - Q[selected_machine])

    print(f"Updated N for {selected_machine}: {N[selected_machine]}")
    print(f"Updated Q for {selected_machine}: {Q[selected_machine]:.2f}")
    print(f"Current Q-values: {Q}")
```

```python
print("\n--- Final Results ---")
print("Participation Count (N):")
for machine, count in N.items():
    print(f"  Machine {machine}: {count} plays")

print("\nEstimated Q-Values (Q):")
for machine, q_value in Q.items():
    print(f"  Machine {machine}: {q_value:.2f}")

print(f"\nTotal reward collected over {N_play} plays: {total_reward_collected:.2f}")
```
